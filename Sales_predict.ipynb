```python
drop_cols = [
    'PHONE',
    'POSTALCODE',
    'CONTACTFIRSTNAME',
    'CONTACTLASTNAME',
    'CUSTOMERNAME',
    'PRODUCTCODE'
]
data.drop(drop_cols,axis=1,inplace=True)
```

```python
data.drop('MONTH_ID',axis=1)
```

```python
data['STATE']=data['STATE'].fillna(data['STATE'].mode()[0])
#data['STATE'] = data['STATE'].fillna(data['STATE'].mode()[0])
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder()
data['STATUS']=oe.fit_transform(data[['STATUS']])
```

```python
data['DEALSIZE']=oe.fit_transform(data[['DEALSIZE']])
data
```

```python
data.info()
```

```python
data.select_dtypes(include='object').columns
```

```python
data=pd.get_dummies(data,columns=['PRODUCTLINE', 'CITY', 'STATE', 'COUNTRY', 'TERRITORY'],drop_first=True)
```

```python
data
```

```python
data.info()
```

```python
data['Day'] = data['ORDERDATE'].dt.day
data['Month'] = data['ORDERDATE'].dt.month
data['Year'] = data['ORDERDATE'].dt.year

data.drop('ORDERDATE', axis=1, inplace=True)
```

```python
bool_cov=data.select_dtypes(include='bool').columns
data[bool_cov] = data[bool_cov].astype(int)
```

```python
data
```

```python
data.info()
```

```python
x=data.drop('SALES',axis=1)
y=data['SALES']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score,rf_

mean_absolute_error(y_test, y_pred)
#r2_score(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
print("MAE :", mean_absolute_error(y_test, y_pred))
print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R²  :", r2_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf.fit(x_train, y_train)

y_pred = rf.predict(x_test)
```

```python
importance = pd.DataFrame({
    'Feature': x.columns,
    'Importance': rf.feature_importances_
})

importance = importance.sort_values(
    by='Importance',
    ascending=False
)

print(importance.head(10))
```

```python
top10 = importance.head(10)

plt.figure(figsize=(8,5))
plt.barh(top10['Feature'], top10['Importance'])
plt.xlabel('Importance')
plt.ylabel('Features')
plt.title('Top 10 Important Features')
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred)
plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Actual vs Predicted Sales")
plt.show()
```

```python

print("MAE :", mean_absolute_error(y_test, y_pred))
print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R²  :", r2_score(y_test, y_pred))
```

```python
print("Train Score:", rf.score(x_train, y_train))
print("Test Score :", rf.score(x_test, y_test))
```
